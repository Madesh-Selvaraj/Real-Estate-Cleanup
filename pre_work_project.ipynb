#### Exercise 09. Are the average prices of "Valdemorillo" and "Galapagar" the same? (★★☆)

Print both average prices and then write a conclusion about them.

```python
ds_valdemorillo = ds[ds["level5"] == "Valdemorillo"]
ds_valdemorillo_mean = ds_valdemorillo["price"].mean()
ds_galapagar = ds[ds["level5"] == "Galapagar"]
ds_galapagar_mean = ds_galapagar["price"].mean()
price_difference= ds_valdemorillo_mean - ds_galapagar_mean
print("The average prices of houses in 'valdemorillo' is:", ds_valdemorillo_mean, "and the average prices of houses in 'galapagar' is:", ds_galapagar_mean, ". In conclusion, Valdermorillo house prices are ", price_difference, "USD higher than Galapagar house prices.")
```

#### Exercise 10. Are the average prices per square meter (price/m2) of "Valdemorillo" and "Galapagar" the same? (★★☆)

Print both average prices and then write a conclusion about it.

Hint: Create a new column called `pps` (price per square meter) and then analyze the values.

```python
import pandas as pd

ds = pd.read_csv('assets/real_estate.csv', sep=';')

ds_valdemorillo = ds[ds["level5"] == "Valdemorillo"]
pps_valdemorillo = (ds_valdemorillo["price"] / ds_valdemorillo["surface"]).mean()
print(pps_valdemorillo)

ds_galapagar = ds[ds["level5"] == "Galapagar"]
pps_galapagar = (ds_galapagar["price"] / ds_galapagar["surface"]).mean()
print(pps_galapagar)

print("The average prices per square meter of Valdemorillo is:", pps_valdemorillo, "and the average prices per square meter of Galapagar is:", pps_galapagar,".", "\nIn conclusion, Galapagar house prices per square meter are ", pps_galapagar - pps_valdemorillo, "USD higher than Valdemorillo house prices per square meter.")

```

#### Exercise 11. Analyze the relation between the surface and the price of the houses (★★☆)

Hint: You can make a `scatter plot`, then write a conclusion about it.

```python
import pandas as pd
import matplotlib.pyplot as plt


ds = pd.read_csv('assets/real_estate.csv', sep=';')
ds_cleaned = ds.dropna(subset=['price', 'surface'])
sc = ds_cleaned.plot.scatter(x='surface', y='price')
sc.set_xlim(15, 500)  # Set the x-axis limits
sc.set_ylim(0, 800000)  # Set the y-axis limits
plt.show()
correlation = ds_cleaned['price'].corr(ds_cleaned['surface'])
print(f"Pearson correlation between 'price' and 'surface': {correlation}", "\nIn conclusion, there is a positive correlation between the price and the surface of the houses in the dataset. This means that as the surface area of a house increases, the price also tends to increase.")
```

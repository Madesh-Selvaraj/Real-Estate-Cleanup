#### Exercise 18. Normalize the variable of prices for each population and plot the 4 histograms in the same plot (you must use the subset obtained in Exercise 14) (★★★)

For the normalization method, you can use the one you consider; there is not a single correct answer to this question. Print the plot and write in the Markdown cell a brief analysis about the plot.

Hint: You can help yourself by reviewing the *multihist* demo of Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

ds = pd.read_csv('assets/real_estate.csv', sep=';')

def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(x - min_val) / (max_val - min_val) for x in data]

ds_Fuenlabrada = ds[ds["level5"] == "Fuenlabrada"]
ds_Fuenlabrada_mean = ds_Fuenlabrada["price"]
Fuenlabrada_normalized = normalize(ds_Fuenlabrada_mean)
ds_Leganés = ds[ds["level5"] == "Leganés"]
ds_Leganés_mean = ds_Leganés["price"]
Leganés_normalized = normalize(ds_Leganés_mean)
ds_Getafe = ds[ds["level5"] == "Getafe"]
ds_Getafe_mean = ds_Getafe["price"]
Getafe_normalized = normalize(ds_Getafe_mean)
ds_Alcorcón = ds[ds["level5"] == "Alcorcón"]
ds_Alcorcón_mean = ds_Alcorcón["price"]
Alcorcón_normalized = normalize(ds_Alcorcón_mean)

plt.hist(Fuenlabrada_normalized, bins=20, alpha=0.9, label='Fuenlabrada', edgecolor='black') 
plt.hist(Leganés_normalized, bins=20, alpha=0.7, label='Leganés', edgecolor='black')
plt.hist(Getafe_normalized, bins=20, alpha=0.5, label='Getafe', edgecolor='black')
plt.hist(Alcorcón_normalized, bins=20, alpha=0.3, label='Alcorcón', edgecolor='black')

plt.xlabel('Population')
plt.ylabel('Frequency')
plt.title('Normalized Price Distribution of Houses in South Belt')
plt.legend()  

plt.show()
```
